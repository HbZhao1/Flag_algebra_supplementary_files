```python
F32 = ThreeGraphTheory.p(5, edges = [[0,1,2],[0,3,4],[1,3,4],[2,3,4]])
F32c = ThreeGraphTheory(5, edges = [[0,1,3],[0,1,4],[0,2,3],[0,2,4],[1,2,3],[1,2,4]])

ThreeGraphTheory.exclude([F32,F32c])
edge = ThreeGraphTheory(3,edges=[[0,1,2]])

print("Number of structures without F32,induced-F32c")
print("size 5: ", len(ThreeGraphTheory.generate(5)))
print("size 6: ", len(ThreeGraphTheory.generate(6)))
```

```python
ThreeGraphTheory.optimize(edge, 6, exact=True, 
                          denom=1024*15*3*5*2, slack_threshold=2e-6,kernel_donom=1024,kernel_threshold=2e-4, file="F32-induced_F32c")
```

```python
ThreeGraphTheory.verify("F32-induced_F32c",edge,6)
```
